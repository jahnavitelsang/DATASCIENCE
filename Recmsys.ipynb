```python
import pandas as pd
import numpy as np
df = pd.read_csv('/content/Movie.csv')
```

```python
df.shape
```

```python
df.head()
```

```python
#Create a user-item matrix
user_df = df.pivot(index='userId',columns='movie',values='rating')
user_df.head()
```

```python
#input nans with 0 values
user_df.fillna(0,inplace=True)   # or i can also use user_df = user_df.fillna(0)
user_df.head()
```

```python
#compute user similarity
from sklearn.metrics import pairwise_distances
from scipy.spatial.distance import cosine,correlation
```

```python
user_sim = 1-pairwise_distances(user_df.values,metric='cosine')
user_sim
```

```python
user_sim.shape
```

```python
np.fill_diagonal(user_sim,0)
user_sim
```

```python
#store sim matrix
user_sim_df = pd.DataFrame(user_sim)
user_sim_df.head()
```

```python
#now set index and coulmns names to user_id\
user_sim_df.index = df.userId.unique()
user_sim_df.columns = df.userId.unique()
user_sim_df.head()
```

```python
user_sim_df.iloc[0:5,0:5]
```

```python
#identify most user similar
user_sim_df.idxmax(axis=1)[0:5]
```
