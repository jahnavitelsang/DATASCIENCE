```python
import pandas as pd
from sklearn.model_selection import train_test_split
```

```python
df = pd.read_csv('/content/Pharma_Industry.csv')
```

```python
X = df.iloc[:,0:3]
y = df['Drug Response']
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)
```

```python
#Define svm with different kernels
from sklearn.svm import SVC
from sklearn.metrics import accuracy_score
```

```python
#Using diff kernels
svm_linear = SVC(kernel='linear')
svm_poly = SVC(kernel='poly', degree = 3)
svm_rbf = SVC(kernel='rbf', gamma = 0.2)
```

```python
#Now lets train the data set
svm_linear.fit(X_train, y_train)
y_pred = svm_linear.predict(X_test)
accuracy_linear = accuracy_score(y_test, y_pred)

```

```python
#Train svm with poly kernel
svm_poly.fit(X_train, y_train)
y_pred = svm_poly.predict(X_test)
accuracy_poly = accuracy_score(y_test, y_pred)
```

```python
svm_rbf.fit(X_train, y_train)
y_pred = svm_rbf.predict(X_test)
accuracy_rbf = accuracy_score(y_test, y_pred)
```

```python
print("Accuracy with linear kernel:", accuracy_linear)
print("Accuracy with poly kernel:", accuracy_poly)
print("Accuracy with rbf kernel:", accuracy_rbf)
```

```python
#Using PCA
from sklearn.decomposition import PCA

pca = PCA(n_components=3)
X_pca = pca.fit_transform(X)
```

```python
X_train_pca, X_test_pca, y_train_pca, y_test_pca = train_test_split(X_pca, y, test_size=0.3, random_state=42)
```

```python
svm_linear.fit(X_pca, y)
y_pred_linear_pca = svm_linear.predict(X_pca)
accuracy_linear_pca = accuracy_score(y, y_pred_linear_pca)
```

```python
svm_poly.fit(X_pca, y)
y_pred_poly_pca = svm_poly.predict(X_pca)
accuracy_poly_pca = accuracy_score(y, y_pred_poly_pca)
```

```python
svm_rbf.fit(X_train_pca, y_train_pca)
y_pred_rbf_pca = svm_rbf.predict(X_pca)
accuracy_rbf_pca = accuracy_score(y, y_pred_rbf_pca)
```

```python
print("Accuracy with linear kernel:", accuracy_linear_pca)
print("Accuracy with poly kernel:", accuracy_poly_pca)
print("Accuracy with rbf kernel:", accuracy_rbf_pca)
```
