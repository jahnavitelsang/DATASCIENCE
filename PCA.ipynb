```python
import numpy as np
import pandas as pd
```

```python
df = pd.read_csv('/content/breast-cancer-wisconsin-data.csv')
```

```python
df.head()
```

```python
df.info()
```

```python
df.describe()
```

```python
#data transformation
from sklearn.preprocessing import LabelEncoder
```

```python
le = LabelEncoder()
```

```python
df['diagnosis'] = le.fit_transform(df['diagnosis'])
```

```python
df['diagnosis']
```

```python
from sklearn.preprocessing import StandardScaler
```

```python
SS = StandardScaler()
```

```python
SS_X = SS.fit_transform(df.iloc[:,2:])
```

```python
SS_X
```

```python
X_new = pd.DataFrame(SS_X)
X_new.columns = df.columns[2:]
```

```python
X_new
```

```python
#pca
from sklearn.decomposition import PCA
```

```python
pca = PCA()
```

```python
pcadata = pca.fit_transform(X_new)
pcadata
```

```python
pcadatas = pd.DataFrame(pcadata)
```

```python
#data particion
X = pcadatas.iloc[:,0:5]
Y = df['diagnosis']
```

```python
#cross validation shuffle split
from sklearn.model_selection import ShuffleSplit, cross_validate, cross_val_score
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import accuracy_score, log_loss
from sklearn.model_selection import train_test_split
```

```python
training_acc = []
test_acc = []
for i in range(1,101):
  X_train, X_test, Y_train, Y_test = train_test_split(X,Y,test_size=0.2, random_state = 42)
  knn = KNeighborsClassifier(n_neighbors=5)
  knn.fit(X_train,Y_train)
  y_pred_train = knn.predict(X_train)
  y_pred_test = knn.predict(X_test)
  training_acc= accuracy_score(Y_train,y_pred_train)
  test_acc = accuracy_score(Y_test,y_pred_test)
```

```python
np.mean(training_acc)
```

```python
np.mean(test_acc)
```
